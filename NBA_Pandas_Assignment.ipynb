```python
import pandas as pd

# DATA UNDERSTANDING

# 1. Load the dataset
df = pd.read_csv("nba.csv")

# 2. Display the first 5 rows
print("Q2: First 5 rows")
display(df.head())

# 3. Check number of rows and columns
print("Q3: Rows and columns")
print(df.shape)

# 4. List all column names
print("Q4: Column names")
print(df.columns.tolist())


# DATA CLEANING

# 5. Convert Salary to numeric (handle errors)
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

# 6. Count missing values in each column
print("Q6: Missing values")
print(df.isnull().sum())

# 7. Drop rows where Salary is missing
df = df.dropna(subset=["Salary"]).copy()

# 8. Fill missing College values with "Unknown"
df["College"] = df["College"].fillna("Unknown")


# FILTERING DATA

# 9. Players from Boston Celtics
print("Q9: Boston Celtics players")
display(df[df["Team"] == "Boston Celtics"])

# 10. Players older than 30
print("Q10: Players older than 30")
display(df[df["Age"] > 30])

# 11. Players who play PG
print("Q11: Point Guards")
display(df[df["Position"] == "PG"])

# 12. Players with Salary greater than 5,000,000
print("Q12: Salary greater than 5 million")
display(df[df["Salary"] > 5000000])


# SORTING

# 13. Sort by Salary (highest first)
print("Q13: Highest salaries first")
display(df.sort_values(by="Salary", ascending=False))

# 14. Sort by Age (youngest first)
print("Q14: Youngest players first")
display(df.sort_values(by="Age", ascending=True))


# GROUPBY OPERATIONS

# 15. Average salary per team
print("Q15: Average salary per team")
display(df.groupby("Team")["Salary"].mean())

# 16. Number of players in each team
print("Q16: Players per team")
display(df.groupby("Team")["Name"].count())

# 17. Maximum salary in each team
print("Q17: Maximum salary per team")
display(df.groupby("Team")["Salary"].max())


# AGGREGATIONS

# 18. Average age
print("Q18: Average age")
print(df["Age"].mean())

# 19. Maximum salary
print("Q19: Maximum salary")
print(df["Salary"].max())

# 20. Minimum weight
print("Q20: Minimum weight")
print(df["Weight"].min())


# COLUMN OPERATIONS

# 21. Create Age_in_5_years
df["Age_in_5_years"] = df["Age"] + 5

print("Q21: Age in 5 years")
display(df[["Name", "Age", "Age_in_5_years"]].head())

# 22. Create Salary_in_Millions
df["Salary_in_Millions"] = df["Salary"] / 1000000

print("Q22: Salary in Millions")
display(df[["Name", "Salary", "Salary_in_Millions"]].head())


# BONUS

# 23. Team with highest total salary payout
print("Q23: Team with highest total salary")
team_salary = df.groupby("Team")["Salary"].sum()
print(team_salary.idxmax())
print(team_salary.max())

# 24. Position with highest average salary
print("Q24: Position with highest average salary")
position_salary = df.groupby("Position")["Salary"].mean()
print(position_salary.idxmax())
print(position_salary.max())
```
