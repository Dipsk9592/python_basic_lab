```python
def calculate_sum(n):

    total = 0

    for number in range(1, n + 1):
        total = total + number

    return total


n = int(input("Enter a positive number: "))

if n > 0:
    result = calculate_sum(n)
    print("Sum:", result)
else:
    print("Please enter a positive number")
```

```python
def count_greater(numbers, value):

    count = 0

    for number in numbers:

        if number > value:
            count = count + 1

    return count


numbers = [10, 25, 40, 55, 70]

value = int(input("Enter comparison value: "))

result = count_greater(numbers, value)

print("Numbers greater than", value, ":", result)
```

```python
def analyze_numbers(numbers):

    count = 0
    total = 0

    even_count = 0
    odd_count = 0

    if len(numbers) == 0:
        return "The list is empty"

    largest = numbers[0]
    smallest = numbers[0]

    for number in numbers:

        count = count + 1

        total = total + number

        if number > largest:
            largest = number

        if number < smallest:
            smallest = number

        if number % 2 == 0:
            even_count = even_count + 1

        else:
            odd_count = odd_count + 1

    average = total / count

    return count, total, average, largest, smallest, even_count, odd_count


numbers = [10, 15, 20, 25, 30]

result = analyze_numbers(numbers)

print("Total values:", result[0])
print("Sum:", result[1])
print("Average:", result[2])
print("Largest:", result[3])
print("Smallest:", result[4])
print("Even numbers:", result[5])
print("Odd numbers:", result[6])
```
