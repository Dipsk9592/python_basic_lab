```python
# Q17 - Build a Small Order Calculator

order_id = input("Enter order ID: ")
city = input("Enter city: ")
unit_price = float(input("Enter unit price: £"))
quantity = int(input("Enter quantity: "))
discount_percent = float(input("Enter discount percentage: "))

subtotal = unit_price * quantity
discount_amount = subtotal * (discount_percent / 100)
final_total = subtotal - discount_amount

print(f"Order {order_id} | {city} | Subtotal: £{subtotal:.2f} | Discount: £{discount_amount:.2f} | Final total: £{final_total:.2f}")
```

```python
# Q18 - Create Validation Flags

valid_quantity = quantity > 0
valid_discount = discount_percent >= 0 and discount_percent <= 100

all_valid = valid_quantity and valid_discount

print(f"Valid quantity: {valid_quantity}")
print(f"Valid discount: {valid_discount}")
print(f"All valid: {all_valid}")
```
